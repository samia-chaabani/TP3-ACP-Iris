```python
import pandas as pd
import numpy as np
```

```python
F = X.drop(["Espece", "Num"], axis=1)
R = F.corr()
print("Matrice de corrélation :")
print(R)
```

```python
valeurs_propres, vecteurs_propres = np.linalg.eig(R)

print("Valeurs propres :")
print(valeurs_propres)

print("Vecteurs propres :")
print(vecteurs_propres)
```

```python

total = np.sum(valeurs_propres)
variance_expliquee = valeurs_propres / total

print("Pourcentage de variance expliquée :")
print(variance_expliquee)
print("Variance cumulée :")
print(np.cumsum(variance_expliquee))
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = X.drop(["Espece","Num"], axis=1)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
pca = PCA()
X_pca = pca.fit_transform(X_scaled)
print("Variance expliquée :", pca.explained_variance_ratio_)
```

```python

index_fleur = X[X["Num"] == "i007"].index[0]
coord = X_pca[index_fleur, 0:2]
print("Coordonnées sur PC1 et PC2 :", coord)
```

```python
pc1 = X_pca[:,0]
contrib = (pc1**2) / np.sum(pc1**2)
max_index = np.argmax(contrib)
print("Fleur qui contribue le plus :", X.iloc[max_index])
```

```python
# Distance totale au centre
dist_total = np.sum(X_scaled**2, axis=1)

# cos² pour tous les individus
cos2 = (X_pca[:,0]**2 + X_pca[:,1]**2) / dist_total

print("Qualité moyenne :", np.mean(cos2))
```

```python
cos2_i027 = cos2[index_fleur]
print("Qualité représentation i027_VIRGINICA :", cos2_i027)
```

```python
# Coordonnées variables
loadings = pca.components_.T * np.sqrt(pca.explained_variance_)

plt.figure(figsize=(6,6))

# Cercle
circle = plt.Circle((0,0), 1, color='blue', fill=False)
plt.gca().add_artist(circle)

# Projection variables
for i, var in enumerate(df.columns):
    plt.arrow(0,0, loadings[i,0], loadings[i,1])
    plt.text(loadings[i,0]*1.1, loadings[i,1]*1.1, var)

plt.axhline(0)
plt.axvline(0)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("Cercle de corrélation")
plt.show()
```
